```python

import numpy as np
import xarray as xr

sea_level_anomaly_data = xr.open_dataset("../../data/sea_level_anomaly_data.nc")
sla_data = sea_level_anomaly_data["sla"].values
data_for_pca = []
print(sla_data.shape)
for i in range(sla_data.shape[1]):
    for j in range(sla_data.shape[2]):
        if not np.isnan(sla_data[:, i, j]).any():
            data_for_pca.append(sla_data[:, i, j])
print(len(data_for_pca))
data_for_pca = np.array(data_for_pca)
print(data_for_pca.shape)

from sklearn.decomposition import PCA

pca = PCA(n_components=10)
pca.fit(data_for_pca)
print(pca.explained_variance_ratio_)
print(sum(pca.explained_variance_ratio_))

pca = PCA(n_components=20)
pca.fit(data_for_pca)
print(pca.explained_variance_ratio_)
print(sum(pca.explained_variance_ratio_))

pca = PCA(n_components=30)
pca.fit(data_for_pca)
print(pca.explained_variance_ratio_)
print(sum(pca.explained_variance_ratio_))

pca = PCA(n_components=40)
pca.fit(data_for_pca)
print(pca.explained_variance_ratio_)
print(sum(pca.explained_variance_ratio_))

pca = PCA(n_components=50)
pca.fit(data_for_pca)
print(pca.explained_variance_ratio_)
print(f" 50 components:{sum(pca.explained_variance_ratio_)}")

pca = PCA(n_components=100)
pca.fit(data_for_pca)
print(pca.explained_variance_ratio_)
print(f" 100 components:{sum(pca.explained_variance_ratio_)}")

pca = PCA(n_components=200)
pca.fit(data_for_pca)
print(pca.explained_variance_ratio_)
print(f" 200 components:{sum(pca.explained_variance_ratio_)}")
```
